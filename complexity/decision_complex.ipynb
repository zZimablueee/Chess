## Methodology for Quantifying Chess Decision Complexity
Based on Barthelemy (2025).

### 1. Engine Evaluation
For a given position at ply $i$, using Stockfish:

- Let $E_1$ be the **best move** evaluation (in centipawns).
- Let $E_2$ be the **second best move** evaluation.
- Define the evaluation gap $\Delta_i$:

$$
\Delta_i = E_1 - E_2
$$

### 2. Probability Model (Boltzmann-like)

We introduce a perception threshold $\Delta_0$ (e.g., $\Delta_0 = 10$ cp for GMs).
The probability of a human finding the optimal move is modeled as:

$$
P(\text{optimal}) = \frac{1}{1 + e^{-\Delta_i / \Delta_0}}
$$

**Limits:**

- If $\Delta_i \gg \Delta_0 \Rightarrow P \to 1$ (Move is obvious).
- If $\Delta_i \ll \Delta_0 \Rightarrow P \to \frac{1}{2}$ (Moves are indistinguishable).

### 3. Information Cost (Decision Difficulty)

The difficulty of choosing the move at ply $i$ is measured in bits:

$$
S(\Delta_i) = -\log_2 P
= \log_2 \left( 1 + e^{-\frac{\Delta_i}{\Delta_0}} \right)
$$

- $\Delta_i \gg \Delta_0 \Rightarrow S \to 0$ bits.
- $\Delta_i \ll \Delta_0 \Rightarrow S \to 1$ bit.

### 4. Cumulative Metrics

For a trajectory of $n$ plies:

$$
S(n) = \sum_{i=1}^{n} S(\Delta_i)
$$

- $S_W(n)$: cumulative difficulty for White.
- $S_B(n)$: cumulative difficulty for Black.

### 5. Asymmetry

$$
A = S_B(n) - S_W(n)
$$

- $A > 0$: Black faces harder decisions.
- $A < 0$: White faces harder decisions.
