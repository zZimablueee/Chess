```python
import chess
def uci_list_string_to_pgn(uci_moves):
    if len(uci_moves) != 1:
        raise ValueError("Expected a list with exactly ONE comma-separated UCI string")
    uci_list = [m.strip() for m in uci_moves[0].split(",") if m.strip()]
    board = chess.Board()
    pgn_parts = []
    move_number = 1
    for ply, uci in enumerate(uci_list, start=1):
        move = chess.Move.from_uci(uci)
        if move not in board.legal_moves:
            raise ValueError(
                f"Illegal move at ply {ply}: {uci}\nFEN: {board.fen()}"
            )
        san = board.san(move)
        if board.turn == chess.WHITE:
            pgn_parts.append(f"{move_number}. {san}")
        else:
            pgn_parts.append(san)
            move_number += 1
        board.push(move)
    return " ".join(pgn_parts)
```
