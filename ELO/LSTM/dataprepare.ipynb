```python
import pandas as pd
import sqlite3
import numpy as np

conn = sqlite3.connect('C:\sqlite3\chess.db')

cols_games = [
    'uid', 'Date', 'White', 'Black', 'WhiteElo', 'BlackElo', 'Result'
]
cols_capture = [
    'uid',
    'WhitePawnAdvanceDepth', 'BlackPawnAdvanceDepth',
    'White_PawnIsolateScoreMean', 'White_PawnOverlapScoreMean', 'White_PawnProtectScoreMean',
    'Black_PawnIsolateScoreMean', 'Black_PawnOverlapScoreMean', 'Black_PawnProtectScoreMean',
    'Black_CaptureRatio', 'White_CaptureRatio',
    'Black_PawnCenter', 'White_PawnCenter',
    'Black_CheckRatio', 'White_CheckRatio',
    'Black_ForceRatio', 'White_ForceRatio'
]
cols_safety = [
    'uid',
    # 白方10项
    'White_CenterControlScoreMean', 'White_PieceActivityScoreMean', 'White_KingSafetyScoreMean', 
    'White_CastlingScoreMean', 'White_KingTropismScoreMean', 'White_KingDefendersScoreMean', 
    'White_KingPawnShieldScoreMean', 'White_KingZoneControlScoreMean', 'White_KingDiagonalExposureScoreMean', 
    'White_KingEscapeSquaresScoreMean',
    # 黑方10项
    'Black_CenterControlScoreMean', 'Black_PieceActivityScoreMean', 'Black_KingSafetyScoreMean', 
    'Black_CastlingScoreMean', 'Black_KingTropismScoreMean', 'Black_KingDefendersScoreMean', 
    'Black_KingPawnShieldScoreMean', 'Black_KingZoneControlScoreMean', 'Black_KingDiagonalExposureScoreMean', 
    'Black_KingEscapeSquaresScoreMean'
]
cols_lifespan=["uid" , "player" , "king_1LifeRatio" ,  "queen_1LifeRatio" ,  "rook_1LifeRatio" , "rook_2LifeRatio" , "bishop_1LifeRatio", "bishop_2LifeRatio" , 
                "knight_1LifeRatio", "knight_2LifeRatio" ,"pawn_1LifeRatio" , "pawn_2LifeRatio" , "pawn_3LifeRatio" , "pawn_4LifeRatio" , "pawn_5LifeRatio" , 
                "pawn_6LifeRatio" , "pawn_7LifeRatio" , "pawn_8LifeRatio", "queen_promo_1LifeRatio" ,"queen_promo_2LifeRatio" , "queen_promo_3LifeRatio" , "queen_promo_4LifeRatio" , 
                "queen_promo_5LifeRatio" , "queen_promo_6LifeRatio" , "queen_promo_7LifeRatio" ,"queen_promo_8LifeRatio" , "rook_promo_1LifeRatio" , "rook_promo_2LifeRatio" ,
                "rook_promo_3LifeRatio" , "rook_promo_4LifeRatio" , "rook_promo_5LifeRatio" ,"rook_promo_6LifeRatio" , "rook_promo_7LifeRatio" , "rook_promo_8LifeRatio" , 
                "bishop_promo_1LifeRatio" , "bishop_promo_2LifeRatio" , "bishop_promo_3LifeRatio" , "bishop_promo_4LifeRatio" , "bishop_promo_5LifeRatio" , 
                "bishop_promo_6LifeRatio" , "bishop_promo_7LifeRatio" , "bishop_promo_8LifeRatio" , "knight_promo_1LifeRatio" , "knight_promo_2LifeRatio" , 
                "knight_promo_3LifeRatio" ,"knight_promo_4LifeRatio" , "knight_promo_5LifeRatio" , 
                "knight_promo_6LifeRatio" , "knight_promo_7LifeRatio" , "knight_promo_8LifeRatio" 
]
#4 tables
def get_sql_cols(cols):
    return ", ".join(cols)
df_games = pd.read_sql(f"SELECT {get_sql_cols(cols_games)} FROM games", conn)
df_capture = pd.read_sql(f"SELECT {get_sql_cols(cols_capture)} FROM CaptureNPawn", conn)
df_safety = pd.read_sql(f"SELECT {get_sql_cols(cols_safety)} FROM kingsafety", conn)
#friendly one
df_lifespan = pd.read_sql(f"SELECT {get_sql_cols(cols_lifespan)} FROM lifespan", conn)
```

```python
for df in [df_games, df_capture, df_safety, df_lifespan]:
    df['uid'] = df['uid'].astype(str).str.strip()

# 转换数值类型 (处理 TEXT 问题)
# 定义需要转数字的列（排除 uid, Date, White, Black 等）
# 除了 games 表里的 metadata，capture 和 safety 表里除了 uid 全是数值
num_cols_capture = [c for c in df_capture.columns if c != 'uid']
num_cols_safety = [c for c in df_safety.columns if c != 'uid']
num_cols_lifespan = [c for c in df_lifespan.columns if c not in ['uid', 'player']]

for col in num_cols_capture:
    df_capture[col] = pd.to_numeric(df_capture[col], errors='coerce')
for col in num_cols_safety:
    df_safety[col] = pd.to_numeric(df_safety[col], errors='coerce')
for col in num_cols_lifespan:
    df_lifespan[col] = pd.to_numeric(df_lifespan[col], errors='coerce')
print("数据类型转换完成。")
```

```python
df_games['uid'] = df_games['uid'].astype(str)
df_capture['uid'] = df_capture['uid'].astype(str)
df_safety['uid'] = df_safety['uid'].astype(str)
df_lifespan['uid'] = df_lifespan['uid'].astype(str)
df_games['uid'] = df_games['uid'].str.strip()
df_capture['uid'] = df_capture['uid'].str.strip()
df_safety['uid'] = df_safety['uid'].str.strip()
df_lifespan['uid'] = df_lifespan['uid'].str.strip()
df_merged = df_games.merge(df_capture, on='uid').merge(df_safety, on='uid')
```

```python
# 看现在的列名到底长什么样
print("清洗前所有列名:", df_merged.columns.tolist())

df_merged.columns = df_merged.columns.str.strip()
df_merged.rename(columns={
    'white': 'White', 
    'black': 'Black', 
    'uid': 'uid',  
    'date': 'Date'
}, inplace=True)

if 'White' not in df_merged.columns:
    print("\n 警告：还是没找到 'White' 列！")
    print("请检查上方打印出的列名，看看代表白方棋手名字的那一列叫什么？")
    # 如果你的列名很怪（比如 'white_player'），请手动重命名：
    # df_merged.rename(columns={'你的怪列名': 'White'}, inplace=True)
else:
    print("\n 成功：找到了 'White' 列，可以继续运行后续代码了！")
```

```python
white_features = [c for c in df_merged.columns if c.startswith('White_')]
black_features = [c for c in df_merged.columns if c.startswith('Black_')]
df_white_perspective = df_merged.copy()
df_white_perspective['player'] = df_white_perspective['White']
df_white_perspective['is_white'] = 1
rename_dict_w = {c: c.replace('White_', 'My_') for c in white_features}
rename_dict_w.update({c: c.replace('Black_', 'Opp_') for c in black_features})
df_white_perspective.rename(columns=rename_dict_w, inplace=True)

df_black_perspective = df_merged.copy()
df_black_perspective['player'] = df_black_perspective['Black']
df_black_perspective['is_white'] = 0
rename_dict_b = {c: c.replace('Black_', 'My_') for c in black_features}
rename_dict_b.update({c: c.replace('White_', 'Opp_') for c in white_features})
df_black_perspective.rename(columns=rename_dict_b, inplace=True)

# df_total的每一行：某一个玩家在某一局棋的表现
df_total = pd.concat([df_white_perspective, df_black_perspective], ignore_index=True)
```

```python
df_total['My_PawnAdvanceDepth'] = np.where(df_total['is_white'] == 1, 
                                           df_total['WhitePawnAdvanceDepth'], 
                                           df_total['BlackPawnAdvanceDepth'])

df_total['Opp_PawnAdvanceDepth'] = np.where(df_total['is_white'] == 1, 
                                            df_total['BlackPawnAdvanceDepth'], 
                                            df_total['WhitePawnAdvanceDepth'])
df_total['My_Elo'] = np.where(df_total['is_white'] == 1, 
                              df_total['WhiteElo'], 
                              df_total['BlackElo'])

df_total['Opp_Elo'] = np.where(df_total['is_white'] == 1, 
                               df_total['BlackElo'], 
                               df_total['WhiteElo'])
cols_to_drop = ['WhitePawnAdvanceDepth', 'BlackPawnAdvanceDepth', 'WhiteElo', 'BlackElo', 'White', 'Black']

df_total.drop(columns=[c for c in cols_to_drop if c in df_total.columns], inplace=True)
df_total.columns
```

```python
df_lifespan['uid']=df_lifespan['uid'].astype(str).str.strip()

# 合并，df_total 是左表 (主表),df_lifespan 是右表 (补充表)
df_final=df_total.merge(df_lifespan, on=['uid', 'player'], how='inner')
# 如果行数变少了很多，说明 lifespan 里缺数据。如果差不多，说明成功。
print(f"合并完成！\ndf_total 行数: {len(df_total)}\ndf_final 行数: {len(df_final)}")
```
